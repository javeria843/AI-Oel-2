```python
#23-AI-12




import collections

graph = {
    'A': [('B', 6), ('C', 2)],
    'B': [('D', 5), ('E', 3)],
    'C': [('F', 4)],
    'D': [('G', 2)],
    'E': [('G', 6)],
    'F': [('G', 1)],
    'G': []
}

def bfs(graph, start, end):
    """Finds a path from start to end using Breadth-First Search."""
    queue = collections.deque([(start, [], 0)])  # (current_node, path_list, total_distance)
    visited = set()

    while queue:
        (current, path, distance) = queue.popleft()

        if current == end:
            return path + [current], distance

        if current not in visited:
            visited.add(current)
            for neighbor, weight in graph.get(current, []):
                queue.append((neighbor, path + [current], distance + weight))

    return None, 0  # No path found

def dfs(graph, start, end, path=[], distance=0, visited=set()):
    """Finds a path from start to end using Depth-First Search."""
    path = path + [start]
    visited.add(start)

    if start == end:
        return path, distance

    for neighbor, weight in graph.get(start, []):
        if neighbor not in visited:
            new_distance = distance + weight
            found_path, total_dist = dfs(graph, neighbor, end, path, new_distance, visited)
            if found_path:
                return found_path, total_dist

    return None, 0  # No path found

#BFS
bfs_path, bfs_distance = bfs(graph, 'A', 'G')
print("BFS Path:", bfs_path)
print("BFS Total Distance:", bfs_distance)

#DFS

dfs_path, dfs_distance = dfs(graph, 'A', 'G', visited=set())
print("DFS Path:", dfs_path)
print("DFS Total Distance:", dfs_distance)
```
